# Tabelas

**1. users (Clientes)**

Contém dados cadastrais e demográficos dos usuários:
* id: Identificador único do usuário (Chave Primária).
* first_name / last_name: Nome e sobrenome.
* email: Endereço de e-mail.
* age: Idade.
* gender: Gênero.
* state / city / country / postal_code: Dados de localização geográfica.
* street_address: Endereço residencial.
* latitude / longitude: Coordenadas geográficas do endereço.
* traffic_source: Canal de marketing pelo qual o usuário chegou (ex: Search, Organic, Facebook).
* created_at: Data e hora de criação da conta.

**2. orders (Pedidos)**

Registra o cabeçalho dos pedidos feitos na plataforma:
* order_id: Identificador único do pedido (Chave Primária).
* user_id: Identificador do usuário que fez a compra (Chave Estrangeira).
* status: Status atual do pedido (ex: Shipped, Complete, Cancelled, Returned).
* gender: Gênero associado ao pedido.
* created_at: Data e hora em que o pedido foi feito.
* returned_at / shipped_at / delivered_at: Datas de retorno, envio e entrega.
* num_of_item: Quantidade total de itens no pedido.

**3. order_items (Itens do Pedido)**

Detelha cada produto individual contido dentro de um pedido:
* id: Identificador único da linha do item (Chave Primária).
* order_id: ID do pedido correspondente (Chave Estrangeira).
* user_id: ID do comprador (Chave Estrangeira).
* product_id: ID do produto comprado (Chave Estrangeira).
* inventory_item_id: ID do item físico específico retirado do estoque (Chave Estrangeira).
* status: Status do item (ex: Processing, Shipped, Complete).
* created_at: Data de criação da linha.
* shipped_at / delivered_at / returned_at: Linha do tempo logística individual do item.
* sale_price: Preço final de venda cobrado pelo produto.

**4. products (Catálogo de Produtos)**

Lista as mercadorias disponíveis para venda na loja de roupas:
* id: Identificador único do produto (Chave Primária).
* cost: Custo de aquisição/fabricação do produto.
* category: Categoria do vestuário (ex: Jeans, Sweaters, Outerwear).
* name: Nome do produto.
* brand: Marca.
* retail_price: Preço sugerido/venda de varejo.
* department: Departamento (ex: Men, Women).
* sku: Código de estoque (Stock Keeping Unit).
* distribution_center_id: Centro de distribuição de onde esse produto é enviado.

**5. inventory_items (Itens de Estoque)**

Controla o estoque físico de produtos nos centros de distribuição:
* id: Identificador único da unidade física em estoque.
* product_id: ID do produto correspondente  (Chave Estrangeira).
* created_at: Quando o item entrou no estoque.
* sold_at: Quando o item foi vendido.
* cost: Custo daquela unidade.
* product_category / product_name / product_brand / product_retail_price / product_department / product_sku: Informações espelhadas do produto para facilidade de consulta.
* product_distribution_center_id: ID do centro de distribuição onde o item está guardado  (Chave Estrangeira).

**6. events (Eventos/Logs Web)**

Registra o comportamento de navegação dos usuários no site ou aplicativo:
* id: Identificador único do evento.
* user_id: ID do usuário (nulo se for um visitante anônimo)  (Chave Estrangeira).
* sequence_number: Ordem do clique dentro da sessão do usuário.
* session_id: Identificador único da sessão de navegação.
* created_at: Data e hora do clique/evento.
* ip_address / city / state / postal_code: Dados geográficos IP da sessão.
* browser: Navegador utilizado (ex: Chrome, Safari).
* traffic_source: Origem do tráfego da sessão.
* uri: O endereço da página da web acessada (ex: /cart, /product/123).
* event_type: Tipo de ação realizada (ex: Home, Department, Product, Cart, Purchase, Cancel).

**7. distribution_centers (Centros de Distribuição)**

Tabela de apoio com as localizações dos armazéns:
* id: Identificador único do centro de distribuição.
* name: Nome da região/cidade do centro.
* latitude / longitude: Localização geográfica do galpão.

# Importação dos dados

```python
import kagglehub
path = kagglehub.dataset_download("mustafakeser4/looker-ecommerce-bigquery-dataset")
```

```python
import os
import pandas as pd

# Carregando todos os datasets de forma prática
df_distribution_centers = pd.read_csv(os.path.join(path, 'distribution_centers.csv'))
df_events = pd.read_csv(os.path.join(path, 'events.csv'))
df_products = pd.read_csv(os.path.join(path, 'products.csv'))
df_users = pd.read_csv(os.path.join(path, 'users.csv'))
df_inventory_items = pd.read_csv(os.path.join(path, 'inventory_items.csv'))
df_orders = pd.read_csv(os.path.join(path, 'orders.csv'))
df_order_items = pd.read_csv(os.path.join(path, 'order_items.csv'))

# Exibindo as primeiras linhas de um exemplo (Users)
print("Dataset de Usuários carregado com sucesso!")
df_users.head()
```

# Tratamento dos dados

```python
import duckdb

query = """
SELECT
  u.id AS user_id,
  u.age,
  u.gender,
  u.country,
  u.state,
  u.traffic_source,
  u.created_at AS user_created_at,
  o.order_id,
  o.status AS order_status,
  o.created_at AS order_created_at,
  o.shipped_at,
  o.delivered_at,
  o.num_of_item AS order_num_of_item,
  oi.id AS order_item_id,
  oi.status AS order_item_status,
  oi.sale_price AS order_item_sale_price,
  COALESCE(p.name, 'Sem Nome') AS product_name,
  p.category AS category,
  p.department AS product_department,
  COALESCE(p.brand, 'Sem Marca') AS brand,
  p.cost AS product_cost,
  p.retail_price AS product_retail_price,
  dc.name AS distribution_center_name,
  ii.id AS item_id,
  ii.created_at AS item_in_sku_at,
  ii.cost AS item_cost
FROM df_order_items oi
INNER JOIN df_orders o ON o.order_id = oi.order_id
LEFT JOIN df_users u ON u.id = oi.user_id
LEFT JOIN df_products p ON p.id = oi.product_id
LEFT JOIN df_inventory_items ii ON ii.id = oi.inventory_item_id
LEFT JOIN df_distribution_centers dc ON dc.id = p.distribution_center_id
WHERE o.status = 'Complete'
  AND o.created_at IS NOT NULL
  AND u.created_at IS NOT NULL
"""

df_result = duckdb.query(query).df()
print(df_result.info())
df_result.head()
```

## Gráficos
**Comparativo de Receita e Margem por Canal (** **traffic\_source** **)**:
* **Tipo de Gráfico**: *Gráfico de Barras Agrupadas*.
* **Por que**: Compara lado a lado o faturamento bruto versus o lucro real de cada canal de aquisição.

**Evolução Mensal do Faturamento por Canal de Origem**:
* **Tipo de Gráfico**: *Gráfico de Linhas*.
* **Por que**: Mostra tendências de crescimento de receita ao longo do tempo (`order_created_at`) separadas por origem.

**LTV e Ticket Médio por Origem de Tráfego**:
* **Tipo de Gráfico**: *Gráfico de Colunas*.
* **Por que**: Evidencia se clientes vindos de canais pagos (ex: Facebook) gastam mais por compra/ao longo da vida do que os de busca orgânica.

**Desempenho por Categoria (`category`), Marca (`brand`) e Departamento (`product_department`)**:
* **Tipo de Gráfico**: *Gráfico de Treemap ou Colunas Horizontais*.
* **Por que**: Identifica rapidamente os produtos campeões de vendas e de maior margem para destacá-los nos anúncios.

**Distribuição Geográfica e Demográfica (`country`, `state`, `gender`, `age`)**:
* **Tipo de Gráfico**: *Mapa Coroplético e Pirâmide Etária / Histograma*.
* **Por que**: Mapeia onde estão os melhores compradores para direcionar a segmentação geográfica e demográfica dos anúncios pagos.
